```python
import spacy
from spacy import displacy
```

## Named Entity Recognition

```python
nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
doc = nlp("Tesla INC is going to acquire twitter for $44 billion")

for ent in doc.ents:
    print("\nText:",ent.text,"| Entity:", ent.label_,"|",spacy.explain(ent.label_))
```

```python
displacy.render(doc, style="ent", jupyter=True)
```

```python
nlp.pipe_labels["ner"]   # List of entities that model can recognize
```

### Using Span

```python
from spacy.tokens import Span
```

```python
doc = nlp("Tesla is going to acquire twitter for $44 billion")

s1 = Span(doc,0,1,label="ORG")
s2 = Span(doc,5,6,label="ORG")

s1 , s2
```

```python
doc.set_ents([s1,s2],default='unmodified')   # Overwrite the existing entities with our custom entities
```

```python
for ent in doc.ents:
    print("\nText:",ent.text,"| Entity:", ent.label_,"|",spacy.explain(ent.label_))
```

```python
displacy.render(doc, style="ent", jupyter=True)
```
