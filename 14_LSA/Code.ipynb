```python
import nltk
import numpy as np
import matplotlib.pyplot as plt
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk import word_tokenize
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import TruncatedSVD
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
titles = [line.rstrip() for line in open('all_book_titles.txt', 'r')]
len(titles)
```

```python
stops = set(stopwords.words('english'))
```

```python
stops = stops.union('intoduction','edition','series','application','approach',
                    'card','access','package','plus','etext',
                    'brief','vol','fundamental','guide','essential','printed',
                    'third','second','fourth','volume')
```

```python
def my_tokenizer(s):

    s = s.lower()   # Lowercasing
    tokens = word_tokenize(s)

    tokens = [t for t in tokens if len(t) > 2] # Remove short words

    tokens = [wordnet_lemmatizer.lemmatize(t) for t in tokens] # Lemmatization

    tokens = [t for t in tokens if t not in stops] # Remove stopwords

    tokens = [t for t in tokens if not any(c.isdigit() for c in t)] # Remove any digits

    return tokens
```

```python
vectorizer = CountVectorizer(tokenizer=my_tokenizer, binary=True)
```

```python
X = vectorizer.fit_transform(titles)   # [Titles x Words] matrix

X.shape
```

```python
index_word_map = vectorizer.get_feature_names_out()
index_word_map
```

```python
X =  X.transpose()  
```

```python
svd = TruncatedSVD(n_components=2)
```

```python
Z = svd.fit_transform(X)  # [Words x Components] matrix

Z.shape
```

```python
(svd.components_).shape
```

```python
import plotly.express as px
```

```python
fig = px.scatter(x=Z[:, 0], y=Z[:, 1], text=index_word_map, size_max=60)
fig.update_traces(textposition='top center')
fig.show()
```
