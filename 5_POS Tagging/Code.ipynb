```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"I will google about facebook")
```

```python
doc.text
```

```python
doc[0].text , doc[0]
```

```python
doc[0] , doc[3] , doc[-1]
```

```python
# Coarse-grained POS Tagging
doc[0].pos_  , doc[1].pos_ , doc[2].pos_ , doc[3].pos_ , doc[4].pos_
```

```python
spacy.explain('PRON') , spacy.explain('AUX') , spacy.explain('ADP') , spacy.explain('PROPN')
```

```python
# Fine-grained POS Tagging
doc[0].tag_  , doc[1].tag_ , doc[2].tag_ , doc[3].tag_ , doc[4].tag_
```

```python
spacy.explain('PRP') , spacy.explain('MD')
```

```python
spacy.explain('VB') , spacy.explain('IN')
```

```python
spacy.explain('NNP')
```

```python
for word in doc:
    print(word.text,"-------->", word.pos_,'|',word.tag_,'|',spacy.explain(word.tag_))
```

```python
doc2 = nlp(u"I left the room")
for word in doc2:
    print(word.text,"------->", word.pos_,'|',word.tag_,'|',spacy.explain(word.tag_))
```

```python
doc3 = nlp(u"to the left of the room")
for word in doc3:
    print(word.text,"------->", word.pos_,'|',word.tag_,'|',spacy.explain(word.tag_))
```

```python
doc4 = nlp(u"I read books on history")
for word in doc4:
    print(word.text,"------->", word.pos_,'|',word.tag_,'|',spacy.explain(word.tag_))
```

```python
doc5 = nlp(u"I have read a book on history")
for word in doc5:
    print(word.text,"------>", word.pos_,'|',word.tag_,'|',spacy.explain(word.tag_))
```
