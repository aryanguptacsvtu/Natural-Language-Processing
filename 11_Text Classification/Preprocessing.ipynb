```python
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences = ['I like eggs and ham.',
             'I love chocolate and cookies.',
             'I hate onions.']
```

```python
MAX_VOCAB_SIZE = 20000
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE)

tokenizer.fit_on_texts(sentences)  # Integer mapping of each token

sequences = tokenizer.texts_to_sequences(sentences)  # Integer Sequence Conversion 
```

```python
print(sequences)
```

```python
# Word to Index mapping
tokenizer.word_index
```

```python
data = pad_sequences(sequences)  # First Method
data 
```

```python
MAX_SEQUENCE_LENGTH = 5

data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH)  # Second Method
data
```

```python
data = pad_sequences(sequences, padding='post')  # Third Method
data 
```

```python
data = pad_sequences(sequences, maxlen = 6)  # Too much padding
data
```

```python
data = pad_sequences(sequences, maxlen = 4)  # Truncation
data
```

```python
data = pad_sequences(sequences, maxlen = 4, truncating='post')  # Truncation
data
```
