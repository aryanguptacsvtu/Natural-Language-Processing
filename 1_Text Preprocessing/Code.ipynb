## Removing HTML Tags

```python
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
text = "<html><body><p> Movie 1</p><p> Actor - Aamir Khan</p><p> Click here to <a href='http://google.com'>download</a></p></body></html>"

remove_html_tags(text)
```

## Removing URLs

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
text1 = 'Check out this notebook https://www.kaggle.com/campusx/notebook8223fc1abb'
text2 = 'Check out my notebook http://www.kaggle.com/campusx/notebook8223fc1abb'
text3 = 'Google search here www.google.com'
text4 = 'For notebook click https://www.kaggle.com/campusx/notebook8223fc1abb to search check www.google.com'
```

```python
print(remove_url(text1))
print(remove_url(text2))
```

```python
print(remove_url(text3))
print(remove_url(text4))
```

## Removing Punctuations

```python
exclude = string.punctuation
exclude
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text
```

```python
text = 'string. With. Punctuation?'

remove_punc(text)  # Slow method
```

```python
def remove_punc_new(text):
    return text.translate(str.maketrans('', '', exclude))
```

```python
text = 's@tring.# Wit%h. Pun?ctuati*on?'

remove_punc_new(text)  # Fast method
```

## Chat word treatment

```python
chat_words = {
    'IMHO' : 'In My Humble Opinion',
    'LOL' : 'Laughing Out Loud',
    'BRB' : 'Be Right Back',
    'IDK' : 'I Don\'t Know',
    'TBH' : 'To Be Honest',
    'BTW':'By The Way',
    'FYI':'For Your Information'
}
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
            
    return " ".join(new_text)
```

```python
chat_conversion('IMHO he is the best')
```

```python
chat_conversion('FYI delhi is the capital of india')
```

## Removing Stopwords

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    
    return " ".join(x)
```

```python
remove_stopwords('probably my all-time favorite movie, a story of selflessness, sacrifice and dedication to a noble cause, but it\'s not preachy or boring. it just never gets old, despite my having seen it some 15 or more times')
```

## Handling Emojis

```python
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

```python
remove_emoji("Lmao 😂😂")
```

## Stemming

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()

def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

```python
text = 'probably my alltime favorite movie a story of selflessness sacrifice and dedication to a noble cause but its not preachy or boring it just never gets old despite my having seen it some 15 or more times in the last 25 years paul lukas performance brings tears to my eyes and bette davis in one of her very few truly sympathetic roles is a delight the kids are as grandma says more like dressedup midgets than children but that only makes them more fun to watch and the mothers slow awakening to whats happening in the world and under her own roof is believable and startling if i had a dozen thumbs theyd all be up for this movie'
stem_words(text)
```

## Lemmatization

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
wl = WordNetLemmatizer()
```

```python
sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"

sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    
    if word in punctuations:
        sentence_words.remove(word)
```

```python
sentence_words
```

```python
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wl.lemmatize(word,pos='v')))
```
